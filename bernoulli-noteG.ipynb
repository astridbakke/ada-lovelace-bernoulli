# Ada Lovelace and the Bernoulli Numbers

This notebook is exploring the historical significance of Ada Lovelace's "Note G". 
Written in the 1840s(!) for Charles Babbage’s mechanical Analytical Engine, this algorithm is widely recognized as the world's first published computer program. 

Rather than just doing simple math, Ada chose to compute the sequence of Bernoulli numbers specifically to demonstrate the complex power and capability of the Analytical Engine. This project translates her historical diagram, essentially a modern executable table, into Python, emulating her original mechanical register operations line-by-line to see how the engine's logic worked under the hood.

```python
# Initialize an array of 25 memory variables (V0 to V24) 
# representing the mechanical columns of discs in the Analytical Engine.
V = [0.0] * 25

# --- 1. SETUP INITIAL VARIABLES ---
V[1] = 1        # Constant 1
V[2] = 2        # Constant 2
V[3] = 4        # n = 4 in Lovelace's program (targeting B8 calculation)

# --- 2. PRE-LOAD BERNOULLI CONSTANTS (Result Variables) ---
V[21] = 1.0 / 6.0      # B1 (representing B2 in modern notation)
V[22] = -1.0 / 30.0    # B3 (representing B4)
V[23] = 1.0 / 42.0     # B5 (representing B6)

# --- 3. FIRST COMPUTATION CYCLE ---
V[4] = V[5] = V[6] = V[2] * V[3]  # Set initial terms to 2 * n (8)
print("Initial Variable State:", V[:7])

V[4] = V[4] - V[1]   # 2n - 1
V[5] = V[5] + V[1]   # 2n + 1

# HISTORICAL NOTE G "BUG": 
# In the original 1843 published table, this division was written backward (V5 / V4). 
# Correcting the code requires fixing it to V4 / V5 to make the math work!
V[11] = V[4] / V[5]   
V[11] = V[11] / V[2]  # Scale by 1/2
V[13] = V[13] - V[11] # Accumulate into result variable V13

V[10] = V[3] - V[1]   # Counter setup (n - 1)
V[7] = V[2] + V[7]    # Shift register value
V[11] = V[6] / V[7]
V[12] = V[21] * V[11] # Multiply by first Bernoulli constant
V[13] = V[12] + V[13] # Add to accumulator
V[10] = V[10] - V[1]  # Decrement counter

print("Result after Loop 1 (V13):", V[13])


# --- 4. SECOND COMPUTATION CYCLE ---
V[6] = V[6] - V[1]
V[7] = V[1] + V[7]
V[8] = V[6] / V[7]
V[11] = V[8] * V[11]

V[6] = V[6] - V[1]
V[7] = V[1] + V[7]
V[9] = V[6] / V[7]
V[11] = V[9] * V[11]

V[12] = V[22] * V[11] # Using B3 constant
V[13] = V[12] + V[13]
V[10] = V[10] - V[1]

print("Result after Loop 2 (V13):", V[13])


# --- 5. THIRD COMPUTATION CYCLE ---
V[6] = V[6] - V[1]
V[7] = V[1] + V[7]
V[8] = V[6] / V[7]
V[11] = V[8] * V[11]

V[6] = V[6] - V[1]
V[7] = V[1] + V[7]
V[9] = V[6] / V[7]
V[11] = V[9] * V[11]

V[12] = V[23] * V[11] # Using B5 constant
V[13] = V[12] + V[13]
V[10] = V[10] - V[1]

print("Final Result in V13:", V[13])
```
